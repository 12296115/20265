```python
#Libraries
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
#Load dataset
dataset_path = "/content/phishing_email_processed.csv"
```

```python
#Basic dataset
final_df = pd.read_csv(dataset_path)

print("Dataset shape:", final_df.shape)
print(final_df["label"].value_counts())
```

```python
#Split
# First: 80% training, 20% temporary
train_df, temp_df = train_test_split(
    final_df,
    test_size=0.20,
    random_state=42,
    stratify=final_df["label"]
)

# Second: split the 20% temporary set equally
# → 10% validation + 10% test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["label"]
)

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

```python
#varify class distribution
print("Train")
print(train_df["Email Type"].value_counts())
print(train_df["label"].value_counts(normalize=True))

print("\nValidation")
print(val_df["Email Type"].value_counts())
print(val_df["label"].value_counts(normalize=True))

print("\nTest")
print(test_df["Email Type"].value_counts())
print(test_df["label"].value_counts(normalize=True))
```

```python
#Save the split
train_path = "/content/phishing_train.csv"
val_path = "/content/phishing_validation.csv"
test_path = "/content/phishing_test.csv"

train_df.to_csv(train_path, index=False, encoding="utf-8")
val_df.to_csv(val_path, index=False, encoding="utf-8")
test_df.to_csv(test_path, index=False, encoding="utf-8")

print("Training set:", train_path)
print("Validation set:", val_path)
print("Test set:", test_path)
```

```python
#Final verification
train_texts = set(train_df["processed_text"])
val_texts = set(val_df["processed_text"])
test_texts = set(test_df["processed_text"])

print("Train ∩ Validation:", len(train_texts & val_texts))
print("Train ∩ Test:", len(train_texts & test_texts))
print("Validation ∩ Test:", len(val_texts & test_texts))
```
